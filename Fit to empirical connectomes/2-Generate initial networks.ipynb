# Purpose
This notebook generates initial networks that match the number of nodes and edges of empirical connectomes.

```python
import pickle
import pandas as pd
import numpy as np

import sys
sys.path.append('..')
from scripts import basic_components as bcomp
```

```python
f = open('./Output/Experimental Adj.pckl', 'rb')
A_real = pickle.load(f)
f.close()
```

```python
K = 20 # number of instances
```

## C. elegans

```python
# network settings
n_vertices = A_real["C.elegans"].shape[0]
edges = np.sum(A_real["C.elegans"]>0)
n_vertices, edges
```

```python
initial_matrix = {}
```

```python
for k in range(K):
    initial_matrix[k] = bcomp. generate_rand_adj(n_vertices, edges, weight_distribution= 'normal', mu=1.0,sig=0.25)

f = open('./Output/inits_CE.pckl', 'wb')
pickle.dump(initial_matrix, f)
f.close()
```
