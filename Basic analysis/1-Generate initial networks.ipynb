# Purpose
This notebook generates initial networks used in the simulations designed to illustrate the behavior of the proposed model.

```python
import numpy as np
import pickle
import sys
sys.path.append("..")
from scripts import basic_components as bcomp
```

# Initial networks for Fig. 4-8 & Supplementary Note 1

```python
# network settings
n_vertices = 100
edges = int(np.round(2 * np.log(n_vertices) * (n_vertices - 1), decimals=0))

# number of different configurations
K = 30
```

```python
initial_matrix = {}
```

```python
for k in range(K):
    initial_matrix[k] = bcomp. generate_rand_adj(n_vertices, edges, weight_distribution= 'normal', mu=1.0, sig=0.25)
```

```python
f = open('./Output/inits.pckl', 'wb')
pickle.dump(initial_matrix, f)
f.close()
```

# Initial networks for Supplementary Note 2

```python
# network settings
n_vertices = 100
density_ls = [0.01, 0.05, 0.2, 0.3]
```

```python
initial_matrix = {}
```

```python
# generate one network for each density
for d in density_ls:
    edges = round(d * n_vertices * (n_vertices - 1))
    initial_matrix[d] = bcomp. generate_rand_adj(n_vertices, edges, weight_distribution= 'normal', mu=1.0,sig=0.25)
```

```python
f = open('./Output/inits_density.pckl', 'wb')
pickle.dump(initial_matrix, f)
f.close()
```
